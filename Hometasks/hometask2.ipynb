```python
raw = [ "T=24.5;P=101.3;H=45", "T=25.1;P=100.8;H=47", "T=26.7;P=100.2;H=52", "T=23.9;P=101.7;H=43", "T=27.2;P=99.8;H=55.1" ]
print(f"""raw: {raw}
""")

T = [float(x.split(";")[0][2:]) for x in raw]
P = [float(x.split(";")[1][2:]) for x in raw]
H = [float(x.split(";")[2][2:]) for x in raw]
print(f"""T: {T}
P: {P}
H: {H}
""")

list_dict_TPH = [{"T" : T[i], "P" : P[i], "H" : H[i]} for i in range(len(T))]
print(f"""Formated list of dicts: {list_dict_TPH}
""")

T_average = sum(T)/len(T)
T_above_average = [x for x in T if x >= T_average]
print(f"""Average T: {T_average}
T above average: {T_above_average}
""")

P_min = min(P)
print(f"""Minimum P: {P_min}
""")

H_int = [x for x in H if x.is_integer()]
print(f"""Int H: {H_int}
""")

print(f"""{'#' + '-' * 30 + '#'}
""")
```

```python
devices = { "D01": {"U": 12, "I": 0.35, "limit": 4.0}, "D02": {"U": 24, "I": 0.18, "limit": 3.5}, "D03": {"U": 8, "I": 0.92, "limit": 5}, "D04": {"U": 9, "I": 0.60, "limit": 5.0} }

devices_results = {key : {"P" : el["U"] * el["I"], "limit" : el["limit"], "is_over" : el["U"] * el["I"] > el["limit"]} for key, el in devices.items()}
devices_over_limits = set([key for key in devices.keys() if devices_results[key]["is_over"]])
P = [el["P"] for el in devices_results.values()]

print(f"""Calculated P for devices: {devices_results}
Devices over limits: {devices_over_limits}
List of P: {P}
Max P device: {max(devices_results, key = lambda device: devices_results[device]["P"])}
""")
```

```python
A = [ [4, 7, 2, 9], [6, 3, 8, 5], [1, 11, 4, 7], [9, 2, 6, 3] ]
A_t = [[A[j][i] for j in range(len(A))] for i in range(len(A))]

main_diag = [A[i][i] for i in range(len(A))]
secondary_diag = [A[i][len(A) - i - 1] for i in range(len(A))]
even_elements = [A[i][j] for i in range(len(A)) for j in range(len(A[i])) if A[i][j] % 2 == 0]

average = sum(A[i][j] for i in range(len(A)) for j in range(len(A[i]))) / (len(A) ** 2)
elements_over_av = [A[i][j] for i in range(len(A)) for j in range(len(A[i])) if A[i][j] > average]

row_sums = {str(i+1) : sum(A[i]) for i in range(len(A))}
column_sums = {str(j+1) : sum(A_t[j]) for j in range(len(A_t))}

max_row = max(row_sums, key=row_sums.get)
max_column = max(column_sums, key=column_sums.get)

print(f"""Main diagonal: {main_diag}
Secondary diagonal: {secondary_diag}
Even elements: {even_elements}

Average: {average}
Elements over average: {elements_over_av}

Row max sum: {max_row}
Column max sum: {max_column}
""")
```

```python
def less_75(number):
    return number < 75

students = { "Мазур": (82, 91, 76), "Бондаренко": (95, 88, 93), "Ткаченко": (67, 74, 81), "Коваленко": (91, 85, 89), "Мельник": (76, 69, 72) }

averages = {key : sum(el) / len (el) for key, el in students.items()}
leader = max(averages, key=averages.get)
good_students = [name for name in students if len(list(filter(less_75, students[name]))) == 0]
unique_results = set([students[name][i] for name in students for i in range(3)])

print(f"""Averages: {averages}
Leader: {leader}
Good students: {good_students}
Unique results: {unique_results}
Quantity of unique results: {len(unique_results)}
""")
```
